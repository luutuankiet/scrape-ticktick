```python
import streamlit as st
import pandas as pd
import duckdb
import os

motherduck_token = os.environ.get("motherduck_token")
con = duckdb.connect(f'md:ticktick_gtd?motherduck_token={motherduck_token}')
cur = con.cursor()

lvl1_lvl2_progress = cur.query("select * from lvl1_lvl2_progress").df()
df=lvl1_lvl2_progress

def highlight_low_val(val):
    if val < 20:
        color = 'red' 
    elif val < 50:
        color = 'yellow'
    elif val < 75:
        color = 'orange'
    else: 
        color = ''
    return f'background-color: {color}'


df=df.style.applymap(highlight_low_val, subset=['done_progress','clarify_progress'])


```
